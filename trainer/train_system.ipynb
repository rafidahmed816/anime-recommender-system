```python
import pandas as pd
```

```python
def combine_features(row):
    return f"{row['Synopsis']} {row['Genres'] if pd.notna(row['Genres']) else ''} {row['Themes'] if pd.notna(row['Themes']) else ''}"
```

```python
df = pd.read_csv("dataset/anime-dataset-2025.csv")
# Fill missing textual fields with empty strings
text_fields = ["Synopsis", "Genres", "Themes", "Studios"]
for field in text_fields:
    if field in df.columns:
        df[field] = df[field].fillna("")
df["features"] = df.apply(combine_features, axis=1)
```

```python
print(df["features"])
```

```python
vectorizer = TfidfVectorizer(stop_words='english', max_features=5000)
anime_matrix = vectorizer.fit_transform(df["features"])
```

```python
print(anime_matrix)
```

```python
# Save
with open("model/vectorizer.pkl", "wb") as f:
    pickle.dump(vectorizer, f)
with open("model/anime_vectors.pkl", "wb") as f:
    pickle.dump(anime_matrix, f)

# df.to_csv("dataset/anime_cleaned.csv", index=False)
```
